```python
raw = mne.io.read_raw_eeglab("../data/raw/sub-001/eeg/sub-001_task-eyesclosed_eeg.set", preload=True)
print(raw)
print(raw.info)
```

```python
raw.plot(n_channels=19, duration=10, scalings=dict(eeg=20e-6))   # more zoomed in
```

```python
raw.plot(n_channels=19, duration=10, scalings=dict(eeg=100e-6))  # more zoomed out
```

```python
raw.plot(n_channels=19, duration=10, start=300, scalings=dict(eeg=20e-6))
```

```python
#NOTE: This creates the bandpass filter (puts floor/ceiling on frequencies being read. 

raw_filtered = raw.copy().filter(l_freq=0.5, h_freq=45.0) 
```

```python
#Running the filtered frequencies into a plot 

raw_filtered.plot(n_channels=19, duration=10, scalings=dict(eeg=20e-6))
```
